# IT Support Ticket Dataset Analysis

This notebook analyzes the support ticket dataset before model development.

```python
# Importing required libraries
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
import re
```

```python
# Loading the dataset
df = pd.read_csv('../dataset/training_data.csv')
df.head()
```

```python
# Finding the number of rows
df.shape[0]
```

```python
# Finding the number of columns
df.shape[1]
```

```python
# Finding the column names
df.columns.tolist()
```

```python
# Checking the data types
df.dtypes
```

```python
# Checking missing values
df.isnull().sum()
```

```python
# Checking duplicate records
df.duplicated().sum()
```

```python
# Finding unique priority classes
df['priority'].unique()
```

```python
# Counting each priority class
priority_counts = df['priority'].value_counts()
priority_counts
```

```python
# Plotting priority distribution
priority_counts.plot(kind='bar')
plt.title('Ticket Priority Distribution')
plt.xlabel('Priority')
plt.ylabel('Number of Tickets')
plt.xticks(rotation=0)
plt.show()
```

```python
# Combining title and description
df['text'] = df['title'].fillna('') + ' ' + df['description'].fillna('')
df[['title', 'description', 'text']].head()
```

```python
# Finding title length
df['title_length'] = df['title'].fillna('').str.len()
df['title_length'].describe()
```

```python
# Finding description length
df['description_length'] = df['description'].fillna('').str.len()
df['description_length'].describe()
```

```python
# Finding common words
text = ' '.join(df['text'].fillna('').str.lower())
words = re.findall(r'\b[a-zA-Z]+\b', text)
word_counts = Counter(words)
word_counts.most_common(20)
```

```python
# Finding common words by priority
for priority in df['priority'].dropna().unique():
    text = ' '.join(df.loc[df['priority'] == priority, 'text'].fillna('').str.lower())
    words = re.findall(r'\b[a-zA-Z]+\b', text)
    print(f'\n{priority}:')
    print(Counter(words).most_common(10))
```

```python
# Comparing ticket length by priority
df.groupby('priority')['description_length'].mean().sort_values(ascending=False)
```

```python
# Plotting description length by priority
df.boxplot(column='description_length', by='priority')
plt.title('Description Length by Priority')
plt.suptitle('')
plt.xlabel('Priority')
plt.ylabel('Description Length')
plt.show()
```

```python
# Displaying sample tickets by priority
for priority in ['Critical', 'High', 'Medium', 'Low']:
    print(f'\n{priority} Tickets')
    display(df[df['priority'] == priority][['title', 'description']].head(3))
```
